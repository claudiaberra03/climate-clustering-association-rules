```python
import numpy as np
from sklearn.metrics import confusion_matrix, adjusted_rand_score
from scipy.optimize import linear_sum_assignment
import pandas as pd
```

```python
# Importare i file per i quali si vuole eseguire il confronto
df1 = pd.read_csv('output/cluster2020_output.csv', encoding='utf-8-sig', sep=';')
df2 = pd.read_csv('output/cluster2035ssp126_t_output.csv',  encoding='utf-8-sig', sep=';')
```

```python
# Funzione per calcolare l'accuratezza considerando le permutazioni delle etichette.
def accuratezza_e_cambiamenti(labels1, labels2):
      
    cm = confusion_matrix(labels1, labels2)
    labels, permutazione = linear_sum_assignment(cm, maximize=True)
    # Crea mappa di permutazione: labels2 -> labels1
    mappa = {col: row for row, col in zip(labels, permutazione)}
    
    # Applica la mappa alle etichette predette
    labels2_allineati = np.array([mappa[label] for label in labels2])
    
    # Calcola accuratezza
    accuratezza = np.mean(labels1 == labels2_allineati)

    # Identifica cambiamenti località per località
    cambiamenti = (labels1 != labels2_allineati)

    return accuratezza, labels2_allineati, cambiamenti
```

```python
# Estraiamo le colonne dal dataframe come numpy array
# Inserire i nomi delle colonne dei clustering da confrontare
labels1 = np.array(df1['kmeans2020_t'])
labels2 = np.array(df2['kmeans20535ssp126_t'])
```

```python
# Calcolo dell'accuratezza mediante la funzione definita
accuratezza, labels2_allineati, cambiamenti  = accuratezza_e_cambiamenti(labels1, labels2)
print(f"Accuratezza (con permutazione): {accuratezza:.2%}")
```
